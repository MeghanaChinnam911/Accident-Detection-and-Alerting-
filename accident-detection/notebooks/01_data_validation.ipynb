# Data Validation
Check that the expected image classes and reference-data files are present before training.

```python
from pathlib import Path

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

expected_paths = {
    "accident images": PROJECT_ROOT / "datasets/accident_images/accident",
    "non-accident images": PROJECT_ROOT / "datasets/accident_images/non_accident",
    "hospital reference CSV": PROJECT_ROOT / "datasets/hospitals/india_hospitals_master_cleaned.csv",
    "police reference CSV": PROJECT_ROOT / "datasets/police/india_police_stations_master_cleaned.csv",
}
for label, path in expected_paths.items():
    status = "found" if path.exists() else "missing"
    print(f"{label}: {status} ({path})")

image_extensions = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}
for class_name in ("accident", "non_accident"):
    folder = PROJECT_ROOT / "datasets/accident_images" / class_name
    images = [p for p in folder.rglob("*") if p.suffix.lower() in image_extensions] if folder.exists() else []
    print(f"{class_name}: {len(images)} images")
```
