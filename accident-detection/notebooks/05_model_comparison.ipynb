# Model Comparison
Load recorded evaluation metrics and compare the completed experiments.

```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
METRICS_FILE = PROJECT_ROOT / "results/metrics.csv"

metrics = pd.read_csv(METRICS_FILE)
if metrics.empty:
    print("No model metrics recorded yet.")
else:
    display(metrics.sort_values("f1_score", ascending=False))
```
